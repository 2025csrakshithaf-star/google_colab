```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import zscore

data={
    "Student": ["A","B","C","D","E","F","G","H","I","J","K","L","M","N","O"],
    "Study_Hours": [2,3,4,5,4,6,5,3,4,5,6,4,5,30,3]
}

df = pd.DataFrame(data)

print("Original Dataset")
display(df)
```

```python
df["Study_Hours"].describe()
```

```python
plt.figure(figsize=(8,4))
sns.boxplot(x=df["Study_Hours"])
plt.title("Study Hours - Box Plot")
plt.xlabel("Study Hours")
plt.show()
```

```python
Q1 = df["Study_Hours"].quantile(0.25)
Q3 = df["Study_Hours"].quantile(0.75)

print("Q1:", Q1)
print("Q3:", Q3)
```

```python
IQR = Q3-Q1
print("IQR:", IQR)
```

```python
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower Bound:", lower_bound)
print("Upper_Bound:", upper_bound)
```

```python
outliers_iqr = df[
    (df["Study_Hours"] < lower_bound) |
    (df["Study_Hours"] > upper_bound)
]

print("IQR Outliers:")
display(outliers_iqr)
```

```python
normal_data_iqr = df[
    (df["Study_Hours"] >= lower_bound) &
    (df["Study_Hours"] <= upper_bound)
]

print("IQR Normal Data:")
display(normal_data_iqr)
```

```python
df["Z_Score"] = zscore(df["Study_Hours"])

print("dataset with zscores")
display(df)
```

```python
z_outlier = df[
    (df["Z_Score"] > 3) |
    (df["Z_Score"] < -3)
]

print("Z score Outliers:")
display(z_outlier)
```

```python
display(outliers_iqr)

display(z_outlier)
```

```python
df_without_outlier = df[
    (df["Study_Hours"] >= lower_bound) &
    (df["Study_Hours"] <= upper_bound)
]

display(df_without_outlier)
```

```python
df_capped = df.copy()

df_capped["Study_Hours"] = df_capped["Study_Hours"].clip(
    lower = lower_bound,
    upper = upper_bound
)

display(df_capped)
```

```python
plt.figure(figsize=(12,4))

plt.subplot(1,2,1)
sns.boxplot(x=df["Study_Hours"])
plt.title("Before Outlier Treatment")

plt.subplot(1,2,2)
sns.boxplot(x=df_capped["Study_Hours"])
plt.title("After Outlier Capping")

plt.show()
```
