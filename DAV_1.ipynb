```python
import numpy as np
import pandas as pd

print("Numpy version:", np.__version__)
print("Pandas version:", pd.__version__)
```

```python
marks = np.array([1,2,3,4])
print("Marks:", marks)
print("Type:",type(marks))
```

```python
print("Number of dimension:", marks.ndim)
print("Shape:",marks.shape)
print("Number of elements:", marks.size)
```

```python
marks = np.array([1,2,3,4])

print("Original:",marks)
print("Add 5:",marks+5)
print("Multiply by 2:",marks*2)
print("Divide by 2:",marks/2)
print("Square:",marks**2)
```

```python
marks = np.array([1,2,3,4,5])

print("Sum:",marks.sum())
print("Mean:",marks.mean())
print("Maximum:",marks.max())
print("Minimum:",marks.min())
```

```python
print("First mark:",marks[0])
print("second:",marks[1])
print("last:",marks[-1])
```

```python
print("Index 1 to 3:",marks[1:4])
print("First three:",marks[:3])
print("last:",marks[2:])
```

```python
stu=np.array([
    [1,2,3],
    [1,2,3],
    [4,5,6]
])

print("Shape:",stu.shape)
```

```python
print("Mean of all values:",stu.mean())
print("Column wise mean:",stu.mean(axis=0))
print("Row wise mean:",stu.mean(axis=1))
```
